```python
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pandas.plotting import scatter_matrix
warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv(r'D:\AI_Projects\ML_StudentScore_Projects\data\raw\StudentsPerformance.csv')
print(data.columns)
print(data.shape)
print(data.head())
print(data.tail())
```

Statistics

```python
data.info()
data.describe(include='all')
```

Find Missing Value/ None Value/ Duplicate Value

```python
print(data.isnull().sum())
print((data.isnull().sum()/data.shape[0])*100)
print((data.duplicated().sum()/data.shape[0])*100)
```

Distribution Data

```python
sns.histplot(data['math score'], kde=True)
plt.title('Math Score Distribution')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.savefig(r'D:\AI_Projects\ML_StudentScore_Projects\reports\figures\math_score_distribution.png')
plt.show()

sns.histplot(data['reading score'], kde=True)
plt.title('Reading Score Distribution')
plt.xlabel('Reading Score')
plt.ylabel('Frequency')
plt.savefig(r'D:\AI_Projects\ML_StudentScore_Projects\reports\figures\reading_score_distribution.png')
plt.show()

sns.histplot(data['writing score'], kde=True)
plt.title('Writing Score Distribution')
plt.xlabel('Writing Score')
plt.ylabel('Frequency')
plt.savefig(r'D:\AI_Projects\ML_StudentScore_Projects\reports\figures\writing_score_distribution.png')
plt.show()
```

Find outliers

```python
data.plot(kind="box", subplots=True, layout=(3,3))
plt.tight_layout()
plt.show()
```

Correlation

```python
corr = data.corr(numeric_only=True)
plt.tight_layout()
sns.heatmap(corr, annot=True)
plt.show()
```

Feature relationship

```python
scatter_matrix(data, figsize=(15, 15))
plt.tight_layout()
plt.show()
```

```python
data[['math score']].value_counts()
sns.countplot(x='math score', data=data)
plt.tight_layout()
# plt.show()
```
